```python
import os
import pandas as pd
import shutil
```

```python
def trimFile(input_folder,output_folder,output_number):
    input_image_arr = []
    for image in os.listdir(input_folder):
        input_image_arr.append(os.path.join(input_folder,image))
    image_df = pd.DataFrame({'input_image_path':input_image_arr})

    output_df = image_df.sample(output_number)

    for i in range(output_df.size):
        in_dir = output_df.iloc[i]['input_image_path']
        out_dir = os.path.join(output_folder,os.path.basename(in_dir))
        out = shutil.copyfile(in_dir,out_dir)
```

```python
# Balance one class by randomly down-sampling it to `output_number` images.
# Point in_path at a class folder of originals; out_path is where the trimmed set goes.
in_path = 'data/raw/ONDE_ONDE'
out_path = 'data/trimmed/ONDE_ONDE'
df = trimFile(in_path, out_path, 400)
```
